# 암 환자와 정상인의 리지스틴 수치

```python
import pandas as pd
dataurl = 'https://raw.githubusercontent.com/Datamanim/datarepo/refs/heads/master/krdatacertificate/s1.csv'
df = pd.read_csv(dataurl)
```

```python
import numpy as np
cls_1 = df[df['Classification']==1]['Resistin']
cls_2 = df[df['Classification']==2]['Resistin']

cls_1_log = np.log(cls_1)
cls_2_log = np.log(cls_2)

cls_1_var = np.var(cls_1_log, ddof=1)
cls_2_var = np.var(cls_2_log, ddof=1)

len(cls_1), len(cls_2)
answer = round(cls_2_var / cls_1_var, 3)
answer = round(cls_2_log.var() / cls_1_log.var(), 3)
```

```python
n1 = len(cls_1)
n2 = len(cls_2)

var_pooled = ((n1-1) * cls_1_var + (n2-1) * cls_2_var) / (n1 + n2 - 2)
answer = round(var_pooled, 3)
```

```python
from scipy import stats
mean1 = cls_1_log.mean()
mean2 = cls_2_log.mean()
t_stat = (mean1-mean2) / np.sqrt(var_pooled * (1/n1 + 1/n2))

p_value = 2*(1-stats.t.cdf(abs(t_stat), df=n1+n2-2))
answer = round(p_value, 3)

ttest_result = stats.ttest_ind(cls_1_log, cls_2_log, equal_var=True)
answer = round(ttest_result.pvalue, 3)
```
