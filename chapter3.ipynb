# 암 환자와 정상인의 리지스틴 수치

```python
import pandas as pd
dataurl = 'https://raw.githubusercontent.com/Datamanim/datarepo/refs/heads/master/krdatacertificate/s1.csv'
df = pd.read_csv(dataurl)
```

```python
import numpy as np
cls_1 = df[df['Classification']==1]['Resistin']
cls_2 = df[df['Classification']==2]['Resistin']

cls_1_log = np.log(cls_1)
cls_2_log = np.log(cls_2)

cls_1_var = np.var(cls_1_log, ddof=1)
cls_2_var = np.var(cls_2_log, ddof=1)

len(cls_1), len(cls_2)
answer = round(cls_2_var / cls_1_var, 3)
answer = round(cls_2_log.var() / cls_1_log.var(), 3)
```

```python
n1 = len(cls_1)
n2 = len(cls_2)

var_pooled = ((n1-1) * cls_1_var + (n2-1) * cls_2_var) / (n1 + n2 - 2)
answer = round(var_pooled, 3)
```

```python
from scipy import stats
mean1 = cls_1_log.mean()
mean2 = cls_2_log.mean()
t_stat = (mean1-mean2) / np.sqrt(var_pooled * (1/n1 + 1/n2))

p_value = 2*(1-stats.t.cdf(abs(t_stat), df=n1+n2-2))
answer = round(p_value, 3)

ttest_result = stats.ttest_ind(cls_1_log, cls_2_log, equal_var=True)
answer = round(ttest_result.pvalue, 3)
```

# 단일 표본 t검정 (one-sample)

```python
# 1. 100명의 키 정보가 들어 있는 데이터가 있다.데이터가 정규성을 만족하는지 확인하라. 그리고 평균키는 165라 판단할수 있는지 귀무가설과 대립가설을 설정한 후 유의수준 5%로 검정하라.
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/height1.csv')
from scipy import stats
stats.shapiro(df)
answer = stats.ttest_1samp(df, 165)
```

```python
# 2. 100명의 키 정보가 들어 있는 데이터가 있다.데이터가 정규성을 만족하는지 확인하라. 그리고 평균키는 165라 판단할수 있는지 귀무가설과 대립가설을 설정한 후 유의수준 5%로 검정하라.
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/height2.csv')
from scipy import stats
stats.shapiro(df)
answer = stats.wilcoxon(df[['height']]-165)
```

# 등분산 검정

```python
# 1. 두개 학급의 시험성적에 대한 데이터이다 그룹간 등분산 검정을 시행하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/scipy2.csv')
from scipy import stats
a = df.loc[df['class']=='A']['score']
b = df.loc[df['class']=='B']['score']
stats.bartlett(a,b)

stats.fligner(a,b, center='median')
stats.fligner(a,b, center='mean')

stats.levene(a,b, center='median')
answer = stats.levene(a,b, center='mean')
```

```python
# 2. 두개 학급의 시험성적에 대한 데이터이다 그룹간 등분산 검정을 시행하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/scipy3.csv')
from scipy import stats
a = df.loc[df['class']=='A']['score']
b = df.loc[df['class']=='B']['score']
stats.bartlett(a,b)

stats.fligner(a,b, center='median')
stats.fligner(a,b, center='mean')

stats.levene(a,b, center='median')
answer = stats.levene(a,b, center='mean')
```

```python
# 3. 두개 학급의 시험성적에 대한 데이터이다 그룹간 등분산 검정을 시행하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/scipy6.csv')
from scipy import stats
stats.bartlett(df['A'], df['B'])

stats.fligner(df['A'], df['B'], center='median')
stats.fligner(df['A'], df['B'], center='mean')

stats.levene(df['A'], df['B'], center='median')
answer = stats.levene(df['A'], df['B'], center='mean')
```
