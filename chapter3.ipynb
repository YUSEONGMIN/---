# 암 환자와 정상인의 리지스틴 수치

```python
import pandas as pd
dataurl = 'https://raw.githubusercontent.com/Datamanim/datarepo/refs/heads/master/krdatacertificate/s1.csv'
df = pd.read_csv(dataurl)
```

```python
import numpy as np
cls_1 = df[df['Classification']==1]['Resistin']
cls_2 = df[df['Classification']==2]['Resistin']

cls_1_log = np.log(cls_1)
cls_2_log = np.log(cls_2)

cls_1_var = np.var(cls_1_log, ddof=1)
cls_2_var = np.var(cls_2_log, ddof=1)

len(cls_1), len(cls_2)
answer = round(cls_2_var / cls_1_var, 3)
answer = round(cls_2_log.var() / cls_1_log.var(), 3)
```

```python
n1 = len(cls_1)
n2 = len(cls_2)

var_pooled = ((n1-1) * cls_1_var + (n2-1) * cls_2_var) / (n1 + n2 - 2)
answer = round(var_pooled, 3)
```

```python
from scipy import stats
mean1 = cls_1_log.mean()
mean2 = cls_2_log.mean()
t_stat = (mean1-mean2) / np.sqrt(var_pooled * (1/n1 + 1/n2))

p_value = 2*(1-stats.t.cdf(abs(t_stat), df=n1+n2-2))
answer = round(p_value, 3)

ttest_result = stats.ttest_ind(cls_1_log, cls_2_log, equal_var=True)
answer = round(ttest_result.pvalue, 3)
```

# 독립표본 검정 (Independent)

```python
# 1. 두개 학급의 시험성적에 대한 데이터이다. 두 학습의 시험 평균(비모수검정의 경우 중위값)은 동일하다 말할 수 있는지 확인 하라
import pandas as pd
df1 = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/ind1.csv')
df2 = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/ind2.csv')
from scipy import stats
stats.shapiro(df1)
stats.shapiro(df2)
# 두집단 정규성 만족

stats.bartlett(df1, df2)
# 등분산 불만족

stats.fligner(df1, df2)
stats.levene(df1, df2)
# 등분산 만족

stats.ttest_ind(df1,df2, equal_var=True)
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 차이가 있다.
```

```python
# 2. 두개 학급의 시험성적에 대한 데이터이다. 두 학습의 시험 평균(비모수검정의 경우 중위값)은 동일하다 말할 수 있는지 확인 하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/scipy5.csv')
group1 = df['A']
group2 = df['B'].dropna()
from scipy import stats
stats.shapiro(group1)
stats.shapiro(group2)
# 정규성 불만족

stats.mannwhitneyu(group1, group2)
stats.ranksums(group1, group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 차이가 없다.
```

```python
# 3. 두개 그룹에 대한 수치형 데이터이다. 두 그룹의 평균은 동일하다 말할 수 있는지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/ind3.csv')

group1 = df[df['group']=='a']['data']
group2 = df[df['group']=='b']['data']
from scipy import stats

stats.shapiro(group1)
stats.shapiro(group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 정규성 만족

stats.bartlett(group1, group2)
stats.fligner(group1, group2)
stats.levene(group1, group2)
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 등분산 아님

stats.ttest_ind(group1, group2, equal_var=False)
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 차이가 있다.
```

```python
# 4. 두개 그룹에 대한 수치형 데이터이다. 두 그룹의 평균은 동일하다 말할 수 있는지 검정하라
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/ind6.csv')

group1 = df['a']
group2 = df['b'].dropna()
from scipy import stats

stats.shapiro(group1)
stats.shapiro(group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 정규성 만족

stats.bartlett(group1, group2)

stats.fligner(group1, group2)
stats.levene(group1, group2)
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 등분산 아님

stats.ttest_ind(group1, group2, equal_var=False)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 차이없다.
```

# 대응표본 t 검정 (paired)

```python
# 1. 특정 질병 집단의 투약 전후의 혈류량 변화를 나타낸 데이터이다. 투약 전후의 변화가 있는지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/rel2.csv')

group1 = df['before']
group2 = df['after']

from scipy import stats
stats.shapiro(group1)
stats.shapiro(group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 정규성 만족

stats.levene(group1, group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 등분산 만족

stats.ttest_rel(group1, group2)
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 차이있다.
```

```python
# 2. 특정 질병 집단의 투약 전후의 혈류량 변화를 나타낸 데이터이다. 투약 전후의 변화가 있는지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/rel3.csv')

group1 = df['before']
group2 = df['after']

from scipy import stats
stats.shapiro(group1)
stats.shapiro(group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 정규성 만족

stats.levene(group1, group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 등분산 만족

stats.ttest_rel(group1, group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 차이없음
```

```python
# 3. 특정 집단의 학습 전후 시험 성적 변화를 나타낸 데이터이다. 시험 전과 후에 차이가 있는지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/rel1.csv')

group1 = df['before']
group2 = df['after']

from scipy import stats
stats.shapiro(group1)
stats.shapiro(group2)
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 정규성 위배

stats.wilcoxon(group1, group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 차이없음
```

```python
# 4. 한 기계 부품의 rpm 수치를 두가지 다른 상황에서 측정했다.(총 70세트) b 상황이 a 상황보다 rpm값이 높다고 말할 수 있는지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/rel4.csv')

group1 = df[df['group']=='a']['rpm']
group2 = df[df['group']=='b']['rpm']
from scipy import stats

stats.shapiro(group1)
stats.shapiro(group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 정규성 만족

stats.levene(group1, group2)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 등분산 만족

stats.ttest_rel(group2, group1, alternative='greater')
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 높다.
```

# 카이제곱 검정 (교차분석)

```python
# 1. 144회 주사위를 던졌을때, 각 눈금별로 나온 횟수를 나타낸다. 이 데이터는 주사위의 분포에서 나올 가능성이 있는지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/dice.csv')
df['expected'] = int(df['counts'].sum()/6)

from scipy import stats
stats.chisquare(df['counts'], df['expected'])
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 동일
```

```python
# 2. 다음 데이터는 어떤 집단의 왼손잡이, 오른손 잡이의 숫자를 나타낸다. 인간의 왼손잡이와 오른손잡이의 비율을 0.2:0.8로 알려져있다.
# 이 집단에서 왼손과 오른손 잡이의 비율이 적합한지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/hands2.csv')

target = df.value_counts().to_frame()
target['expected'] = [int(target['count'].sum()*0.8), int(target['count'].sum()*0.2)]

from scipy import stats
stats.chisquare(target['count'], target['expected'])
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 다르다.
```

```python
# 3. 다음 데이터는 국민 기초체력을 조사한 데이터이다. 성별과 등급이 독립적인지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/body/body.csv')
cdf = pd.crosstab(df['측정회원성별'], df['등급'])

from scipy import stats
result = stats.chi2_contingency(cdf)
result.pvalue
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 독립적이지 않다.
```

```python
# 4. 성별에 따른 동아리 활동 참석 비율을 나타낸 데이터이다. 성별과 참석간에 관련이 있는지 검정하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/fe2.csv',index_col=0)
target = df.iloc[:2,:2]

from scipy import stats
result = stats.chi2_contingency(target)
result.pvalue
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 독립적이다.

# 4보다 작은 셀이 20% 넘어서 피셔 정확검정
result = stats.fisher_exact(target)
result.pvalue
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 독립적이지 않다.
```

# 세 집단 이상 검정 (independent)

```python
# 1. x1,x2,x3의 변수들의 평균의 차이가 존재하는지 검정하라. 차이가 존재한다면 사후 분석까지 진행하라
import pandas as pd 
df = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/scipy/anova.csv')

data = df.x1
data2 = df.x2
data3 = df.x3

from scipy import stats
stats.shapiro(data)
stats.shapiro(data2)
stats.shapiro(data3)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 정규성 만족

stats.levene(data,data2,data3)
# p-value가 0.05보다 크므로 귀무가설 기각못함 -> 등분산 만족

# from statsmodels.formula.api import ols
# from statsmodels.stats.anova import anova_lm
dfm = df.melt()
# model = ols('value ~ C(variable)', dfm).fit()
# anova_lm(model)

from scipy import stats
stats.f_oneway(data, data2, data3)
# p-value가 0.05보다 작으므로 귀무가설 기각 -> 집단간 차이가 있다.

from statsmodels.stats.multicomp import pairwise_tukeyhsd
posthoc = pairwise_tukeyhsd(dfm.value, dfm.variable, alpha=0.05)
print(posthoc)
```
