# 03 Grouping

```python
# 44. 데이터를 로드하고 상위 5개 컬럼을 출력하라
DataUrl = 'https://raw.githubusercontent.com/Datamanim/pandas/main/AB_NYC_2019.csv'
df = pd.read_csv(DataUrl)
Ans = df.head(5)
```

```python
# 45. 데이터의 각 host_name의 빈도수를 구하고 host_name으로 정렬하여 상위 5개를 출력하라
Ans = df.groupby('host_name').size()
Ans = df['host_name'].value_counts().sort_index()
# Ans.head(5)
```

```python
# 46. 데이터의 각 host_name의 빈도수를 구하고 빈도수 기준 내림차순 정렬한 데이터 프레임을 만들어라. 빈도수 컬럼은 counts로 명명하라
Ans = df['host_name'].value_counts().sort_values(ascending=False).reset_index(name='counts')
Ans = df.groupby('host_name').size().to_frame().rename(columns={0:'counts'}).sort_values('counts', ascending=False)
# Ans.head(5)
```

```python
# 47. neighbourhood_group의 값에 따른 neighbourhood컬럼 값의 갯수를 구하여라
Ans = df.groupby(['neighbourhood_group','neighbourhood'], as_index=False).size()
```

```python
# 48. neighbourhood_group의 값에 따른 neighbourhood컬럼 값 중 neighbourhood_group그룹의 최댓값들을 출력하라
Ans = df.groupby(['neighbourhood_group','neighbourhood'], as_index=False).size().groupby(['neighbourhood_group'], as_index=False).max()
```

```python
# 49. neighbourhood_group 값에 따른 price값의 평균, 분산, 최대, 최소 값을 구하여라
Ans = df[['neighbourhood_group','price']].groupby('neighbourhood_group').agg(['mean','var','max','min'])
```

```python
# 50. neighbourhood_group 값에 따른 reviews_per_month 평균, 분산, 최대, 최소 값을 구하여라
Ans = df[['neighbourhood_group','reviews_per_month']].groupby(['neighbourhood_group']).agg(['mean','var','max','min'])
```

```python
# 51. neighbourhood 값과 neighbourhood_group 값에 따른 price 의 평균을 구하라
Ans = df.groupby(['neighbourhood','neighbourhood_group'])['price'].mean()
```

```python
# 52. neighbourhood 값과 neighbourhood_group 값에 따른 price 의 평균을 계층적 indexing 없이 구하라
Ans = df.groupby(['neighbourhood','neighbourhood_group'])['price'].mean().unstack()
```

```python
# 53. neighbourhood 값과 neighbourhood_group 값에 따른 price 의 평균을 계층적 indexing 없이 구하고 nan 값은 -999값으로 채워라
Ans = df.groupby(['neighbourhood','neighbourhood_group'])['price'].mean().unstack(fill_value=-999)
Ans = df.groupby(['neighbourhood','neighbourhood_group'])['price'].mean().unstack().fillna(-999)
```

```python
# 54. 데이터중 neighbourhood_group 값이 Queens값을 가지는 데이터들 중 neighbourhood 그룹별로 price값의 평균, 분산, 최대, 최소값을 구하라
Ans = df.loc[df['neighbourhood_group'].str.contains('Queens')].groupby('neighbourhood')['price'].agg(['mean','var','max','min'])
Ans = df.loc[df['neighbourhood_group']=='Queens'].groupby('neighbourhood')['price'].agg(['mean','var','max','min'])
```

```python
# 55. 데이터중 neighbourhood_group 값에 따른 room_type 컬럼의 숫자를 구하고 neighbourhood_group 값을 기준으로 각 값의 비율을 구하여라
Ans = df[['neighbourhood_group','room_type']].groupby(['neighbourhood_group','room_type']).size().unstack().astype('float')
Ans.loc[:,:] = Ans.values / Ans.sum(axis=1).to_numpy().reshape(-1,1)
Ans
```

# 04 Apply, Map

```python
# 56. 데이터를 로드하고 데이터 행과 열의 갯수를 출력하라
DataUrl = 'https://raw.githubusercontent.com/Datamanim/pandas/main/BankChurnersUp.csv'
df = pd.read_csv(DataUrl, index_col=0)
Ans = df.shape
```

```python
# 57. Income_Category의 카테고리를 map 함수를 이용하여 다음과 같이 변경하여 newIncome 컬럼에 매핑하라
dic = {
    'Unknown'        : 'N',
    'Less than $40K' : 'a',
    '$40K - $60K'    : 'b',
    '$60K - $80K'    : 'c',
    '$80K - $120K'   : 'd',
    '$120K +'        : 'e'
}
df['newIncome'] = df['Income_Category'].map(lambda x: dic[x])
Ans = df['newIncome']
```

```python
# 58. Income_Category의 카테고리를 apply 함수를 이용하여 다음과 같이 변경하여 newIncome 컬럼에 매핑하라
def changeCategory(x):
    if   x =='Unknown':
        return 'N'
    elif x =='Less than $40K':
        return 'a'
    elif x =='$40K - $60K':   
        return 'b'
    elif x =='$60K - $80K':    
        return 'c'
    elif x =='$80K - $120K':   
        return 'd'
    elif x =='$120K +' :     
        return 'e'
df['newIncome'] = df['Income_Category'].apply(changeCategory)
Ans = df['newIncome']
```

```python
# 59. Customer_Age의 값을 이용하여 나이 구간을 AgeState 컬럼으로 정의하라. (0~9: 0, 10~19: 10, 20~29: 20 …) 각 구간의 빈도수를 출력하라
def changeAge(x):
    if x <= 9:
        return '0'
    elif x <= 19:
        return '10'
    elif x <= 29:
        return '20'
    elif x <= 39:
        return '30'
    elif x <= 49:
        return '40'    
    elif x <= 59:
        return '50'
    elif x <= 69:
        return '60'
    elif x <= 79:
        return '70'
df['AgeState'] = df['Customer_Age'].apply(changeAge)
Ans = df['AgeState'].value_counts().sort_index()

df['AgeState'] = df['Customer_Age'].map(lambda x: x//10 * 10)
Ans = df['AgeState'].value_counts().sort_index()
```

```python
# 60. Education_Level의 값중 Graduate 단어가 포함되는 값은 1 그렇지 않은 경우에는 0으로 변경하여 newEduLevel 컬럼을 정의하고 빈도수를 출력하라
def changeEduLevel(x):
    if 'Graduate' in x:
        return 1
    else:
        return 0
df['newEduLevel'] = df['Education_Level'].apply(changeEduLevel)
Ans = df['newEduLevel'].value_counts()

df['newEduLevel'] = df['Education_Level'].map(lambda x: 1 if 'Graduate' in x else 0)
Ans = df['newEduLevel'].value_counts()
```

```python
# 61. Credit_Limit 컬럼값이 4500 이상인 경우 1 그외의 경우에는 모두 0으로 하는 newLimit 정의하라. newLimit 각 값들의 빈도수를 출력하라
def changeLimit(x):
    if x >= 4500:
        return 1
    else:
        return 0
df['newLimit'] = df['Credit_Limit'].apply(changeLimit)
df['newLimit'].value_counts()

df['newLimit'] = df['Credit_Limit'].map(lambda x: 1 if x >= 4500 else 0)
df['newLimit'].value_counts()
```

```python
# 62. Marital_Status 컬럼값이 Married 이고 Card_Category 컬럼의 값이 Platinum인 경우 1 그외의 경우에는 모두 0으로 하는 newState컬럼을 정의하라. newState의 각 값들의 빈도수를 출력하라
def changeState(x):
    if x['Marital_Status'] == 'Married' and x['Card_Category'] == 'Platinum':
        return 1
    else:
        return 0
df['newState'] = df.apply(changeState, axis=1)
Ans = df['newState'].value_counts()
```

```python
# 63. Gender 컬럼값 M인 경우 male F인 경우 female로 값을 변경하여 Gender 컬럼에 새롭게 정의하라. 각 value의 빈도를 출력하라
def changeGender(x):
    if x == 'M':
        return 'male'
    elif x == 'F':
        return 'female'
    else:
        return '0'
df['newGender'] = df['Gender'].apply(changeGender)
Ans = df['newGender'].value_counts()

df['newGender'] = df['Gender'].map(lambda x: 'male' if x == 'M' else 'female' if x == 'F' else '0')
Ans = df['newGender'].value_counts()
```

```python
# 64. 데이터를 로드하고 각 열의 데이터 타입을 파악하라
DataUrl = 'https://raw.githubusercontent.com/Datamanim/pandas/main/timeTest.csv'
df = pd.read_csv(DataUrl)
# df.info()
```

```python
# 65. Yr_Mo_Dy을 판다스에서 인식할 수 있는 datetime64타입으로 변경하라
df['Yr_Mo_Dy'] = df['Yr_Mo_Dy'].astype('datetime64[ns]')
df['Yr_Mo_Dy'] = pd.to_datetime(df['Yr_Mo_Dy'])
```

```python
# 66. Yr_Mo_Dy에 존재하는 년도의 유일값을 모두 출력하라
Ans = df['Yr_Mo_Dy'].dt.year.unique()
```

```python
# 67. Yr_Mo_Dy에 년도가 2061년 이상의 경우에는 모두 잘못된 데이터이다. 해당경우의 값은 100을 빼서 새롭게 날짜를 Yr_Mo_Dy 컬럼에 정의하라
def fix_century(x):
    year = x.year - 100 if x.year >= 2061 else x.year
    # return pd.to_datetime('%s-%s-%s' % (year, x.month, x.day))
    # return pd.to_datetime('{}-{}-{}'.format(year, x.month, x.day))
    # return pd.to_datetime('{year}-{month}-{day}'.format(year=year, month=x.month, day=x.day))
    return pd.to_datetime(f'{year}-{x.month}-{x.day}')
df['Yr_Mo_Dy'] = df['Yr_Mo_Dy'].apply(fix_century)
```

```python
# 68. 년도별 각컬럼의 평균값을 구하여라
Ans = df.groupby(df['Yr_Mo_Dy'].dt.year).mean().iloc[:,1:]
```

```python
# 69. weekday 컬럼을 만들고 요일별로 매핑하라 ( 월요일: 0 ~ 일요일 :6)
df['weekday'] = df['Yr_Mo_Dy'].dt.weekday
Ans = df['weekday'].to_frame()
```

```python
# 70. weekday컬럼을 기준으로 주말이면 1 평일이면 0의 값을 가지는 WeekCheck 컬럼을 만들어라
def checkWeek(x):
    if x == 5 or x == 6:
        return 1
    else:
        return 0
df['WeekCheck'] = df['weekday'].apply(checkWeek)
df['WeekCheck'] = df['weekday'].map(lambda x: 1 if x == 5 or x == 6 else 0)
df['WeekCheck'] = df['weekday'].map(lambda x: 1 if x in [5,6] else 0)
Ans = df['WeekCheck'].to_frame()
```

```python
# 71. 년도, 일자 상관없이 모든 컬럼의 각 달의 평균을 구하여라
Ans = df.groupby([df['Yr_Mo_Dy'].dt.month]).mean().iloc[:,1:]
```

```python
# 72. 모든 결측치는 컬럼기준 직전의 값으로 대체하고 첫번째 행에 결측치가 있을경우 뒤에있는 값으로 대채하라
df = df.ffill().bfill()
```

```python
# 73. 년도 - 월을 기준으로 모든 컬럼의 평균값을 구하여라
Ans = df.groupby([df['Yr_Mo_Dy'].dt.year, df['Yr_Mo_Dy'].dt.month]).mean().iloc[:,1:]
Ans = df.groupby([df['Yr_Mo_Dy'].dt.to_period('M')]).mean().iloc[:,1:]
```

```python
# 76. 년-월-일:시 컬럼을 pandas에서 인식할 수 있는 datetime 형태로 변경하라. 서울시의 제공데이터의 경우 0시가 24시로 표현된다
DataUrl = 'https://raw.githubusercontent.com/Datamanim/pandas/main/seoul_pm.csv'
df = pd.read_csv(DataUrl)

def changeDate(x):
    import datetime
    date, hour = x.split(':')
    if hour == '24':
        hour = '00'
        FinalDate = pd.to_datetime(date + ' ' + hour) + datetime.timedelta(days=1)
    else:
        FinalDate = pd.to_datetime(date + ' ' + hour)
    return FinalDate
df['(년-월-일:시)'] = df['(년-월-일:시)'].apply(changeDate)
```
